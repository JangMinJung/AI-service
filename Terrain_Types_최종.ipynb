```python
from owlready2 import *

onto_tank_1 = get_ontology("http://test.org/onto.owl")
```

```python
with onto_tank_1:
    
    # 도메인 클래스
    class Terrain(Thing):pass

    # 경사도에 따른 지형 구분
    class LowSlope(Terrain): pass
    class MiddleSlope(Terrain): pass
    class HighSlope(Terrain): pass
    AllDisjoint([LowSlope, MiddleSlope, HighSlope])

    # 강에 따른 지형 구분
    class ShallowStreams(Terrain): pass
    class MiddleStreams(Terrain): pass
    class DeepStreams(Terrain): pass
    AllDisjoint([ShallowStreams, MiddleStreams, DeepStreams])
   
    # 산림에 따른 지형 구분
    class SparseVegetation(Terrain): pass
    class DenseVegetation(Terrain): pass
    class SeverelyDenseVegetation(Terrain):pass
    AllDisjoint([SparseVegetation, DenseVegetation, SeverelyDenseVegetation])
 
    # 평균 속도 datatypeproperty
    class hasTypicalseeds(DataProperty,FunctionalProperty):
        domain = [Thing]
        range = [float]
    
    # 결과(레인지)클래스
    class Unrestricted(Terrain): 
        equivalent_to = [hasTypicalseeds.value(24)]
    class Restricted(Terrain): 
        equivalent_to = [hasTypicalseeds.value(16)]
    class Severely_restricted(Terrain): 
        equivalent_to = [hasTypicalseeds.value(1)]
    AllDisjoint([Restricted, Severely_restricted, Unrestricted])
    
    # Datatypeproperty 정의
    class hasSlope(DataProperty,FunctionalProperty): 
        domain = [Terrain]
        range = [float]

    class hasCurrent(DataProperty,FunctionalProperty):
        domain = [Terrain]
        range = [float]

    class hasDepth(DataProperty,FunctionalProperty):
        domain = [Terrain]
        range = [float]

    class hasWidth(DataProperty,FunctionalProperty):
        domain = [Terrain]
        range = [float]
    
    class hasSpacing(DataProperty,FunctionalProperty):
        domain = [Terrain]
        range = [float]
    
    class hasTrunk(DataProperty,FunctionalProperty):
        domain = [Terrain]
        range = [float]
```

```python

def terrain_type(onto):
    # 탱크의 너비
    tank_1_avlblength = 50

    for Terrain in onto.classes():
        for terrain in Terrain.instances():
            if terrain.hasSlope <30:
                terrain.is_a.append(LowSlope)
            elif terrain.hasSlope < 45 and terrain.hasSlope >= 30:
                terrain.is_a.append(MiddleSlope)
            elif terrain.hasSlope>=45:
                terrain.is_a.append(HighSlope)

        for terrain in Terrain.instances():
            if terrain.hasDepth <2 and terrain.hasWidth <5:
                terrain.is_a.append(ShallowStreams)
            elif terrain.hasDepth>=2 and terrain.hasDepth<4 and terrain.hasCurrent <5 and terrain.hasWidth <tank_1_avlblength:
                terrain.is_a.append(MiddleStreams)
            elif terrain.hasDepth>=4 and terrain.hasCurrent>=5 and terrain.hasWidth>=tank_1_avlblength:
                terrain.is_a.append(DeepStreams)

        for terrain in Terrain.instances():
            if terrain.hasSpacing >=20 and terrain.hasTrunk <2:
                terrain.is_a.append(SparseVegetation)
            elif terrain.hasSpacing <20 and terrain.hasTrunk >=2 and terrain.hasTrunk <6:
                terrain.is_a.append(DenseVegetation)
            elif terrain.hasSpacing <20 and terrain.hasTrunk > 6:
                terrain.is_a.append(SeverelyDenseVegetation)
        # print(terrain.is_a)

def terrain_restriction(onto):
    for Terrain in onto.classes():
        for terrain in Terrain.instances():
            
            # terrain이 LowSlope, ShallowStreams, SparseVegetation의 인스턴스인 경우
            if isinstance(terrain, LowSlope) and isinstance(terrain, ShallowStreams) and isinstance(terrain, SparseVegetation):
                terrain.is_a.append(Unrestricted)
                
            # terrain이 HighSlope, DeepStreams, SeverelyDenseVegetation 중 하나의 인스턴스인 경우
            elif isinstance(terrain, (HighSlope, DeepStreams, SeverelyDenseVegetation)):
                terrain.is_a.append(Severely_restricted)
                
            # 위 조건에 해당하지 않는 경우 기본적으로 Restricted 클래스로 설정
            else:
                terrain.is_a.append(Restricted)
        # print(terrain.is_a)
                
def tank_seed(onto):
    for Terrain in onto.classes():
        for terrain in Terrain.instances():
            if isinstance(terrain,Unrestricted):
                terrain.hasTypicalseeds = 24
            elif isinstance(terrain, Restricted):
                terrain.hasTypicalseeds = 16
            elif isinstance(terrain, Severely_restricted):
                terrain.hasTypicalseeds = 1
        # print(terrain.is_a)         
```

```python
terrain_1 = Terrain("terrain_1")
terrain_1.hasSlope = 15
terrain_1.hasDepth = 1
terrain_1.hasWidth = 3
terrain_1.hasSpacing = 30
terrain_1.hasTrunk = 1


terrain_2 = Terrain("terrain_2")
terrain_2.hasSlope = 50
terrain_2.hasDepth = 1
terrain_2.hasWidth = 3
terrain_2.hasSpacing = 30
terrain_2.hasTrunk = 1    
```

```python
terrain_type(onto_tank_1)
```

```python
terrain_restriction(onto_tank_1)
```

```python
tank_seed(onto_tank_1)
```
